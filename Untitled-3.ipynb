```python
def filter_strings(words, k):
    result = []

    for word in words:
        if len(word) > k:
            result.append(word)

    return result


words = ["apple", "pie", "banana"]
k = 3

print(filter_strings(words, k))
```

```python
def area(shape, x, y=0):

    if shape == "circle":
        result = 3.14 * x * x
        print("Circle Area:", result)

    elif shape == "rectangle":
        result = x * y
        print("Rectangle Area:", result)


area("circle", 5)
area("rectangle", 4, 6)
```

```python
class Restaurant:

    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, name, table):
        reservation = {
            "name": name,
            "table": table
        }

        self.book_table.append(reservation)

    def customer_order(self, name, order):
        customer = {
            "name": name,
            "order": order
        }

        self.customer_orders.append(customer)


restaurant = Restaurant()

# Add items to menu
restaurant.add_item_to_menu("Pizza", 100)
restaurant.add_item_to_menu("Burger", 80)
restaurant.add_item_to_menu("Pasta", 90)

# Make table reservations
restaurant.book_tables("Ahmed", 5)
restaurant.book_tables("Ali", 2)

# Customer orders
restaurant.customer_order("Ahmed", "Pizza")
restaurant.customer_order("Ali", "Burger")

# Print menu
print("Menu:")
print(restaurant.menu_items)

# Print table reservations
print("Table Reservations:")
print(restaurant.book_table)

# Print customer orders
print("Customer Orders:")
print(restaurant.customer_orders)
```

```python
class Customer:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_age(self):
        return self.age

    def set_age(self, age):
        self.age = age


class Bike:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_price(self):
        return self.price

    def set_price(self, price):
        self.price = price


class MainBikeRental:

    def __init__(self):
        self.bikes = []
        self.rentals = []

    def get_bikes(self):
        return self.bikes

    def set_bikes(self, bikes):
        self.bikes = bikes

    def get_rentals(self):
        return self.rentals

    def set_rentals(self, rentals):
        self.rentals = rentals

    def requestBike(self, customer, number_of_bikes):

        if customer.get_age() <= 6:
            print("Customer must be over 6 years old")
            return

        if number_of_bikes <= len(self.bikes):

            for i in range(number_of_bikes):
                self.bikes.pop()

            self.rentals.append(number_of_bikes)

            print("Bike rented successfully")

        else:
            print("Wait 10 mins")

    def returnBike(self, rentalTime, num_of_bikes):

        total = rentalTime * 40

        for i in range(num_of_bikes):
            self.bikes.append("Bike")

        print("Total bill:", total)

    def addBike(self, bike):
        self.bikes.append(bike)

    def totalCost(self):

        total = 0

        for bike in self.bikes:
            if isinstance(bike, Bike):
                total = total + bike.get_price()

        return total


# EL Maaaaaiiiiiin 


# Create customer
customer = Customer("Ahmed", 20)

print("Customer Name:", customer.get_name())
print("Customer Age:", customer.get_age())


# Create bikes
bike1 = Bike("Bike 1", 500)
bike2 = Bike("Bike 2", 600)
bike3 = Bike("Bike 3", 700)


# Create bike rental shop
shop = MainBikeRental()


# Add bikes to shop
shop.addBike(bike1)
shop.addBike(bike2)
shop.addBike(bike3)

print("Total cost of bikes:", shop.totalCost())


# Request bikes
shop.requestBike(customer, 2)


# Show number of bikes left
print("Bikes left:", len(shop.get_bikes()))


# Return bikes
shop.returnBike(3, 2)


# Show number of bikes after returning
print("Bikes after return:", len(shop.get_bikes()))
```
